## Bước 3. Phân Tích Đơn Biến & Phân Phối (Univariate Analysis & Distributions)

* **Phân phối lớp nhãn (Class Distribution):** Đánh giá tính cân bằng của nhãn mục tiêu `label`.
* **Phân phối độ dài email:** Phân tích độ dài văn bản của email (số từ và số ký tự).
* **Tần suất từ khóa phổ biến:** Tìm các từ khóa xuất hiện nhiều nhất trong toàn bộ tập dữ liệu.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")

# 3a. Phân phối lớp nhãn mục tiêu (Class Distribution)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Biểu đồ cột
label_counts = data['label'].value_counts()
axes[0].bar(['Not Spam (0)', 'Spam (1)'], label_counts.values, color=['#2196F3', '#FF5722'])
axes[0].set_title('Số lượng email theo nhãn')
axes[0].set_ylabel('Số lượng')
for i, v in enumerate(label_counts.values):
    axes[0].text(i, v + len(data)*0.01, str(v), ha='center', fontweight='bold')

# Biểu đồ hình tròn
axes[1].pie(label_counts.values, labels=['Not Spam (0)', 'Spam (1)'], autopct='%1.1f%%',
            colors=['#2196F3', '#FF5722'], startangle=90, explode=(0, 0.05))
axes[1].set_title('Tỷ lệ phân phối nhãn')

plt.suptitle('Class Distribution - Biến mục tiêu (label)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

ratio = label_counts[1] / label_counts[0] if label_counts[0] > 0 else 0
print(f"\nTỷ lệ Spam (1) / Not Spam (0) = {ratio:.2f} : 1")
print(f"→ Nhận xét: Dữ liệu phân bố cực kỳ cân bằng và đồng đều (Tỷ lệ xấp xỉ 55% vs 45%).")
```

```python
# 3b. Phân tích phân phối độ dài của email (Email Length Distribution)
# Điền giá trị rỗng cho các email không có nội dung body
data['body'] = data['body'].fillna('')
data['subject'] = data['subject'].fillna('')

# Tính độ dài số từ (Word Count) và số ký tự (Char Count)
data['body_len_words'] = data['body'].fillna('').apply(lambda x: len(x.split()))
data['body_len_chars'] = data['body'].fillna('').apply(len)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Phân phối số từ trong email body (giới hạn dưới 1000 từ để biểu đồ rõ ràng)
axes[0].hist(data[data['body_len_words'] < 1000]['body_len_words'], bins=50, color='#42A5F5', edgecolor='white')
axes[0].set_title('Phân phối số từ trong Email Body (< 1000 từ)')
axes[0].set_xlabel('Số từ')
axes[0].set_ylabel('Số email')
skew_words = data['body_len_words'].skew()
axes[0].text(0.95, 0.95, f'Skew={skew_words:.2f}', transform=axes[0].transAxes,
             ha='right', va='top', fontsize=10, color='red', fontweight='bold')

# Phân phối số ký tự trong email body (giới hạn dưới 5000 ký tự)
axes[1].hist(data[data['body_len_chars'] < 5000]['body_len_chars'], bins=50, color='#26A69A', edgecolor='white')
axes[1].set_title('Phân phối số ký tự trong Email Body (< 5000 ký tự)')
axes[1].set_xlabel('Số ký tự')
axes[1].set_ylabel('Số email')
skew_chars = data['body_len_chars'].skew()
axes[1].text(0.95, 0.95, f'Skew={skew_chars:.2f}', transform=axes[1].transAxes,
             ha='right', va='top', fontsize=10, color='red', fontweight='bold')

plt.suptitle('Phân phối độ dài Email Body', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 3c. Phân tích tần suất từ khóa xuất hiện nhiều nhất (Top Word Frequencies)
from sklearn.feature_extraction.text import CountVectorizer

# Trích xuất top 42 từ khóa phổ biến nhất toàn bộ tập dữ liệu (loại bỏ stop words tiếng Anh)
vectorizer = CountVectorizer(stop_words='english', max_features=42)
word_counts = vectorizer.fit_transform(data['body'])
sum_words = word_counts.sum(axis=0)

words_freq = [(word, sum_words[0, idx]) for word, idx in vectorizer.vocabulary_.items()]
words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
words, word_counts_list = zip(*words_freq)

plt.figure(figsize=(10, 5))
plt.bar(words, word_counts_list, color='purple')
plt.title('Top 42 từ khóa xuất hiện nhiều nhất trong tập dữ liệu', fontsize=13, fontweight='bold')
plt.xlabel('Từ khóa')
plt.ylabel('Tần suất xuất hiện')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## Bước 4. Phân Tích Song Biến & Tương Quan (Bivariate Analysis, Correlation & Patterns)

* **So sánh độ dài:** So sánh độ dài trung bình của email giữa 2 lớp nhãn mục tiêu.
* **Box plot phân phối:** Vẽ biểu đồ hộp so sánh chi tiết phân phối độ dài email theo lớp.
* **Từ khóa đặc trưng từng lớp:** So sánh các từ khóa xuất hiện nhiều nhất trong thư thường và thư rác.
* **Phân tích dấu hiệu đặc trưng (Patterns):**  đặc trưng phishing (chứa từ khóa crime, attack, hoặc virus trong body)

```python
# 4a. So sánh độ dài trung bình của email giữa Spam (1) vs Not Spam (0)
mean_len = data.groupby('label')[['body_len_words', 'body_len_chars']].mean()
print("Độ dài trung bình của Email theo nhãn mục tiêu:")
print(mean_len)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(['Not Spam (0)', 'Spam (1)'], mean_len['body_len_words'], color=['#2196F3', '#FF5722'])
axes[0].set_title('Số từ trung bình trong Email')
axes[0].set_ylabel('Số từ')

axes[1].bar(['Not Spam (0)', 'Spam (1)'], mean_len['body_len_chars'], color=['#2196F3', '#FF5722'])
axes[1].set_title('Số ký tự trung bình trong Email')
axes[1].set_ylabel('Số ký tự')

plt.suptitle('So sánh độ dài email trung bình: Spam vs Not Spam', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 4b. Box plot so sánh phân phối độ dài email theo nhãn
plt.figure(figsize=(8, 6))
# Giới hạn dưới 1000 từ để biểu đồ trực quan, không bị bóp méo hộp
sns.boxplot(data=data[data['body_len_words'] < 1000], x='label', y='body_len_words',
            hue='label', palette={0: '#2196F3', 1: '#FF5722', '0': '#2196F3', '1': '#FF5722'},
            legend=False)
plt.title('Biểu đồ hộp (Box Plot) số từ trong email theo nhãn (< 1000 từ)', fontsize=13, fontweight='bold')
plt.xticks([0, 1], ['Not Spam (0)', 'Spam (1)'])
plt.xlabel('Prediction (label)')
plt.ylabel('Số từ')
plt.tight_layout()
plt.show()
# Lọc dữ liệu dưới 1000 từ giống như lúc vẽ biểu đồ
data_filtered = data[data['body_len_words'] < 1000]

# Xem chi tiết các mốc tứ phân vị (25%, 50%, 75%) của từng nhóm nhãn (label)
print(data_filtered.groupby('label')['body_len_words'].describe())
```

```python
# 4c. So sánh Top 42 từ khóa phổ biến nhất ở mỗi lớp
vec_spam = CountVectorizer(stop_words='english', max_features=42)
spam_words = vec_spam.fit_transform(data[data['label'] == 1]['body']).sum(axis=0)
spam_freq = sorted([(w, spam_words[0, idx]) for w, idx in vec_spam.vocabulary_.items()], key=lambda x: x[1], reverse=True)

vec_ham = CountVectorizer(stop_words='english', max_features=42)
ham_words = vec_ham.fit_transform(data[data['label'] == 0]['body']).sum(axis=0)
ham_freq = sorted([(w, ham_words[0, idx]) for w, idx in vec_ham.vocabulary_.items()], key=lambda x: x[1], reverse=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Trực quan cho lớp thường (Not Spam)
ham_w, ham_c = zip(*ham_freq)
axes[0].bar(ham_w, ham_c, color='#2196F3')
axes[0].set_title('Top 10 từ khóa phổ biến trong Not Spam (0)')
axes[0].set_xticks(range(len(ham_w)))
axes[0].set_xticklabels(ham_w, rotation=90)

# Trực quan cho lớp rác (Spam)
spam_w, spam_c = zip(*spam_freq)
axes[1].bar(spam_w, spam_c, color='#FF5722')
axes[1].set_title('Top 10 từ khóa phổ biến trong Spam (1)')
axes[1].set_xticks(range(len(spam_w)))
axes[1].set_xticklabels(spam_w, rotation=90)

plt.suptitle('So sánh các từ khóa xuất hiện nhiều nhất giữa 2 lớp', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 4d. Phân tích đặc trưng đặc thù: Từ đe dọa (phishing)
import pandas as pd
import matplotlib.pyplot as plt

# 1. Tạo đặc trưng has_phishing (chứa từ khóa crime, attack, hoặc virus trong body)
data['has_phishing'] = data['body'].apply(lambda x: 'crime' in str(x) or 'attack' in str(x) or 'virus' in str(x))

# 2. Tính tỷ lệ trung bình (%) theo nhãn
rates_phishing = data.groupby('label')['has_phishing'].mean() * 100

print("Tỷ lệ % email chứa từ đe dọa theo nhãn:")
print(rates_phishing)

# 3. Vẽ biểu đồ so sánh tỷ lệ giữa Not Spam và Spam
fig, ax = plt.subplots(figsize=(6, 5))

# Vẽ biểu đồ cột (Bar plot)
ax.bar(['Not Spam (0)', 'Spam (1)'], rates_phishing.values, color=['#2196F3', '#FF5722'], width=0.5)

ax.set_title('Tỷ lệ email chứa các từ đe dọa (crime/attack/virus)', fontsize=12, fontweight='bold', pad=15)
ax.set_ylabel('Tỷ lệ (%)', fontsize=10)
ax.set_ylim(0, max(rates_phishing.values) + 3)

# Ghi nhãn giá trị phần trăm cụ thể lên đầu mỗi cột
for i, v in enumerate(rates_phishing.values):
    ax.text(i, v + 0.5, f"{v:.2f}%", ha='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()
```

## Bước 5. Nhận Diện Dữ Liệu Thiếu, Trùng Lặp & Ngoại Lệ (Missing Data, Duplicate Data & Outliers)

* **Dữ liệu thiếu:** Kiểm tra các giá trị khuyết (`NaN`) trong toàn bộ tập dữ liệu.
* **Dữ liệu trùng lặp:** Tính toán số dòng trùng lặp dựa trên nội dung thư (`subject` + `body`).
* **Giá trị ngoại lệ (Outliers):** Sử dụng phương pháp toán học IQR để lọc ra các email có độ dài quá biến động bất thường.

```python
# 5a. Kiểm tra dữ liệu thiếu (Missing Data)
missing_count = data.isnull().sum()
print("=== KIỂM TRA DỮ LIỆU THIẾU ===")
print(f"Tổng số giá trị khuyết (NaN) trong dataset: {missing_count.sum()}")
print(missing_count)
```

```python
# 5b. Kiểm tra dữ liệu trùng lặp (Duplicate Data)
dup_all = data.duplicated().sum()
dup_content = data.duplicated(subset=['subject', 'body']).sum()

print("=== KIỂM TRA DỮ LIỆU TRÙNG LẶP ===")
print(f"Số dòng trùng lặp trên tất cả các cột: {dup_all}")
print(f"Số dòng trùng lặp trên nội dung (Subject + Body): {dup_content}")
```

```python
# 5c. Phát hiện giá trị ngoại lệ (Outliers) bằng phương pháp IQR trên thuộc tính độ dài email
print("=== KIỂM TRA GIÁ TRỊ NGOẠI LỆ (OUTLIERS) ===")

Q1 = data['body_len_words'].quantile(0.25)
Q3 = data['body_len_words'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data['body_len_words'] < lower) | (data['body_len_words'] > upper)]
n_outliers = len(outliers)

print(f"Các thông số của phân bố độ dài từ trong Email Body:")
print(f"  - Q1 (25%): {Q1} từ")
print(f"  - Q3 (75%): {Q3} từ")
print(f"  - IQR: {IQR} từ")
print(f"  - Ngưỡng biên dưới (Lower bound): {lower} từ")
print(f"  - Ngưỡng biên trên (Upper bound): {upper} từ")
print(f"  - Số dòng có độ dài ngoại lệ: {n_outliers} / {len(data)} ({n_outliers/len(data)*100:.2f}%) ")
```

```python
# 5d. Trực quan hóa giá trị ngoại lệ (Outliers)
plt.figure(figsize=(10, 4))
sns.boxplot(x=data['body_len_words'], color='#FF7043')
plt.title('Biểu đồ hộp (Box Plot) thể hiện phân phối số từ và các điểm ngoại lệ', fontsize=13, fontweight='bold')
plt.xlabel('Số từ trong Email')
plt.tight_layout()
plt.show()
```

```python
# 6b. Kết luận & Ghi chú cho bước Feature Engineering tiếp theo
import matplotlib.pyplot as plt
import seaborn as sns

print("="*60)
print("    KẾT LUẬN EDA & GHI CHÚ CHO FEATURE ENGINEERING")
print("="*60)
print("""
1. MISSING DATA (Giá trị khuyết thiếu):
   → Phát hiện chính xác 462 giá trị khuyết thiếu (NaN) ở cột 'receiver' (người nhận).
   → Cột 'subject' và 'body' không có dòng trống nào.
   → Khuyến nghị: Điền khuyết cột 'receiver' bằng nhãn mới 'Unknown' ở bước Feature Engineering.

2. DUPLICATE DATA (Dữ liệu trùng lặp):
   → Kết quả quét thực tế cho thấy có 0 dòng trùng lặp nội dung (Subject + Body).
   → Khuyến nghị: Bỏ qua bước xóa trùng lặp (drop_duplicates) để tối ưu hiệu năng.

3. CLASS DISTRIBUTION (Phân phối lớp):
   → Nhãn mục tiêu 'label' cực kỳ cân bằng: 55.8% Spam (21,842 dòng) và 44.2% Ham (17,312 dòng).
   → Khuyến nghị: Không cần áp dụng các phương pháp cân bằng dữ liệu.

4. FEATURE ENGINEERING ĐỀ XUẤT:
   → Chuyển đổi văn bản tự do sang ma trận số bằng phương pháp TF-IDF.
   → Tận dụng các đặc trưng hành vi và thời gian đắc lực đã được chứng minh qua EDA:
     * 'has_phishing' (Dấu hiệu chứa các từ đe dọa: crime, attack, virus).
""")
print("="*60)

# --- PHẦN TRỰC QUAN HÓA KẾT LUẬN (DATA QUALITY VISUALIZATION) ---
metrics = ['Missing Values (NaN)', 'Duplicate Rows', 'Outliers (Email Length > 565 words)']
values = [462, 0, 2478]

plt.figure(figsize=(10, 5))
# Sử dụng bảng màu: Vàng cho cảnh báo thiếu, Xanh lá cho an toàn (0 trùng), Đỏ cho outliers cần nén
colors = ['#FFC107', '#4CAF50', '#FF5722'] 
bars = plt.barh(metrics, values, color=colors, height=0.5)

# Hiển thị số lượng dòng cụ thể trên đầu mỗi thanh cột
for bar in bars:
    width = bar.get_width()
    plt.text(width + 50, bar.get_y() + bar.get_height()/2, f'{int(width)} dòng', 
             va='center', ha='left', fontweight='bold', fontsize=10)

plt.title('Tóm tắt các vấn đề về Chất lượng Dữ liệu (Data Quality Issues)', fontsize=13, fontweight='bold')
plt.xlabel('Số lượng dòng bị ảnh hưởng')
plt.xlim(0, 3000)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
